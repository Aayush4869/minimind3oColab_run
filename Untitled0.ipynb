```python

%cd /content/minimind-o

file_path = "eval_omni.py"

with open(file_path, "r", encoding="utf-8") as f:
    code = f.read()

start_marker = "    if '0' in modes:\n"
end_marker = "    if '1' in modes:\n"

start = code.index(start_marker)
end = code.index(end_marker, start)

interactive_block = '''    if '0' in modes:
        print("\\n\\n==================== Interactive MiniMind-3O ====================")
        idx = 0

        while True:
            try:
                prompt = input("\\n📝 Enter your prompt (or type 'exit' to quit): ").strip()
            except (EOFError, KeyboardInterrupt):
                print("\\nExiting chat.")
                break

            if prompt.lower() in ("exit", "quit"):
                print("Exiting chat.")
                break

            if not prompt:
                continue

            eval_sample(
                model, tokenizer, args, idx, prompt, None,
                f"text-{idx:02d}.mp3"
            )
            idx += 1

'''

code = code[:start] + interactive_block + code[end:]

with open(file_path, "w", encoding="utf-8") as f:
    f.write(code)

print("Interactive prompt loop installed!")
```
